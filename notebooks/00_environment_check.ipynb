# Phase 0: Environment Check

Verify PySpark, H3, Delta Lake work.

```python
# PySpark version
import pyspark
print(f"PySpark: {pyspark.__version__}")
```

```python
# SparkSession with Delta Lake
from pyspark.sql import SparkSession
from delta import configure_spark_with_delta_pip

builder = SparkSession.builder \
    .appName("AeroPulse3D") \
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension") \
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")

spark = configure_spark_with_delta_pip(builder).getOrCreate()
print(f"Spark UI: http://localhost:4040")
spark
```

```python
# H3 spatial indexing
import h3
print(f"H3: {h3.__version__}")

# Test: Delhi coords to H3 cell
delhi_lat, delhi_lng = 28.6139, 77.2090
h3_cell = h3.latlng_to_cell(delhi_lat, delhi_lng, 7)
print(f"Delhi H3 (res 7): {h3_cell}")
```

```python
# Delta Lake write/read test
from pyspark.sql import Row

test_data = [Row(id=1, value="test")]
df = spark.createDataFrame(test_data)

delta_path = "/home/jovyan/work/data/delta/test"
df.write.format("delta").mode("overwrite").save(delta_path)

df_read = spark.read.format("delta").load(delta_path)
df_read.show()
print("Delta Lake: OK")
```
